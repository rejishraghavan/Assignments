```python
# Assignment 6 - each question is followed immediately by its solution.

# Helper used by the project-search and funding questions below.
def iter_projects(system):
    for university_name, university in system["universities"].items():
        for faculty_name, faculty in university.get("faculties", {}).items():
            for department_name, department in faculty.get("departments", {}).items():
                for lab_name, lab in department.get("labs", {}).items():
                    for project_name, project in lab.get("research_projects", {}).items():
                        yield university_name, faculty_name, department_name, lab_name, project_name, project

# 1. Print the location of GTU_India.
print("Q1:", university_system["universities"]["GTU_India"]["location"])

# 2. Print the dean of the engineering faculty in GTU_Singapore.
print("Q2:", university_system["universities"]["GTU_Singapore"]["faculties"]["engineering"]["dean"])

# 3. Print the HOD of the computer_science department.
india_cs = university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]
print("Q3:", india_cs["hod"])

# 4. Print all labs under computer_science.
print("Q4:", list(india_cs["labs"].keys()))

# 5. Print the funding amount of the DeepVision project.
print("Q5:", india_cs["labs"]["AI Lab"]["research_projects"]["DeepVision"]["funding"])

# 6. Print all research project names in GTU_India.
india_projects = [(lab_name, project_name) for uni, _, _, lab_name, project_name, _ in iter_projects(university_system) if uni == "GTU_India"]
print("Q6:", india_projects)

# 7. Print all ongoing projects in GTU_India.
ongoing_india = [project_name for uni, _, _, _, project_name, project in iter_projects(university_system) if uni == "GTU_India" and project.get("status") == "ongoing"]
print("Q7:", ongoing_india)

# 8. Calculate the total funding for all projects in AI Lab.
ai_projects = india_cs["labs"]["AI Lab"]["research_projects"]
print("Q8:", sum(project["funding"] for project in ai_projects.values()))

# 9. Print all researchers working under Systems Lab.
systems_projects = india_cs["labs"]["Systems Lab"]["research_projects"].values()
print("Q9:", [researcher for project in systems_projects for researcher in project["researchers"]])

# 10. Count students enrolled in all computer_science courses.
print("Q10:", sum(course["students_enrolled"] for course in india_cs["courses"].values()))

# 11. Print all university names in the system.
print("Q11:", list(university_system["universities"].keys()))

# 12. Print all faculty names under GTU_India.
print("Q12:", list(university_system["universities"]["GTU_India"]["faculties"].keys()))

# 13. Print all department names across all universities.
departments = [department_name for university in university_system["universities"].values() for faculty in university.get("faculties", {}).values() for department_name in faculty.get("departments", {})]
print("Q13:", departments)

# 14. Print all deans in the entire system.
print("Q14:", [faculty["dean"] for university in university_system["universities"].values() for faculty in university.get("faculties", {}).values() if "dean" in faculty])

# 15. Print all project names across both universities.
print("Q15:", [project_name for _, _, _, _, project_name, _ in iter_projects(university_system)])

# 16. Add a new research project under AI Lab.
ai_projects["SmartHealth"] = {"funding": 50000, "researchers": ["Ananya", "Vikram"], "status": "ongoing"}
print("Q16:", ai_projects["SmartHealth"])

# 17. Change DeepVision's status to completed.
ai_projects["DeepVision"]["status"] = "completed"
print("Q17:", ai_projects["DeepVision"]["status"])

# 18. Increase CloudScale funding by 50,000.
cloudscale = india_cs["labs"]["Systems Lab"]["research_projects"]["CloudScale"]
cloudscale["funding"] += 50000
print("Q18:", cloudscale["funding"])

# 19. Add a new university to the system.
university_system["universities"]["GTU_UAE"] = {"location": "Dubai", "ranking": 4, "faculties": {}}
print("Q19:", university_system["universities"]["GTU_UAE"])

# 20. Add a course under the finance department.
finance = university_system["universities"]["GTU_India"]["faculties"]["management"]["departments"]["finance"]
finance["courses"]["Financial Analytics"] = {"credits": 3, "students_enrolled": 60}
print("Q20:", finance["courses"]["Financial Analytics"])

# 21. Find the project with the highest funding.
highest_project = max(iter_projects(university_system), key=lambda entry: entry[5]["funding"])
print("Q21:", highest_project[4], highest_project[5]["funding"])

# 22. Find total funding across all universities.
print("Q22:", sum(project["funding"] for *_, project in iter_projects(university_system)))

# 23. Find the university with the best (lowest-numbered) ranking.
best_university = min(university_system["universities"].items(), key=lambda entry: entry[1]["ranking"])
print("Q23:", best_university[0], best_university[1]["ranking"])

# 24. Print all completed projects.
print("Q24:", [project_name for _, _, _, _, project_name, project in iter_projects(university_system) if project.get("status") == "completed"])

# 25. Count unique researchers in GTU_India.
india_researchers = {researcher for uni, _, _, _, _, project in iter_projects(university_system) if uni == "GTU_India" for researcher in project["researchers"]}
print("Q25:", len(india_researchers))

# 26. Find the lab with the maximum total project funding.
lab_funding = {}
for _, _, _, lab_name, project_name, project in iter_projects(university_system):
    lab_funding.setdefault(lab_name, 0)
    lab_funding[lab_name] += project["funding"]
print("Q26:", max(lab_funding.items(), key=lambda item: item[1]))

# 27. Print projects grouped by status.
projects_by_status = {}
for _, _, _, _, project_name, project in iter_projects(university_system):
    projects_by_status.setdefault(project["status"], []).append(project_name)
print("Q27:", projects_by_status)

# 28. Print all courses with more than 100 students enrolled.
courses_over_100 = [(university_name, course_name, course["students_enrolled"]) for university_name, university in university_system["universities"].items() for faculty in university.get("faculties", {}).values() for department in faculty.get("departments", {}).values() for course_name, course in department.get("courses", {}).items() if course["students_enrolled"] > 100]
print("Q28:", courses_over_100)

# 29. Find the total number of faculties in the entire system.
print("Q29:", sum(len(university.get("faculties", {})) for university in university_system["universities"].values()))

# 30. Search for a project name and print its full details.
def search_project(project_name, system=university_system):
    for university_name, faculty_name, department_name, lab_name, name, details in iter_projects(system):
        if name == project_name:
            result = {"university": university_name, "faculty": faculty_name, "department": department_name, "lab": lab_name, "project": details}
            print(result)
            return result
    print(f"Project not found: {project_name}")
    return None

search_project("DeepVision")
```
