```python
# Load the provided employee JSON file from the assignment folder.
import json
from pathlib import Path

json_path = Path("employees.json")
if not json_path.exists():
    json_path = Path("/home/raghavanr/rejish/BATCH/Assign/Day 3/employees.json")
with json_path.open("r") as file:
    employees = json.load(file)

print(f"Loaded {len(employees)} employees from {json_path.name}.")
```

```python
# Assignment questions: Employee JSON database, Q1-Q29
# The original employees dictionary is kept unchanged; modification exercises use copies.

# Q1. Check whether Employee ID 1050 exists.
print("Q1:", "1050" in employees)

# Q2. Ask the user to enter an Employee ID and display whether it exists.
def employee_exists(employee_id):
    return str(employee_id) in employees
# Interactive use: entered_id = input("Enter an Employee ID: ").strip()
entered_id = "1050"  # Sample input for a non-interactive notebook run.
print("Q2:", "Employee exists" if employee_exists(entered_id) else "Employee not found")

# Q3. If Employee ID 1050 exists, display all employee details.
if "1050" in employees:
    print("Q3:", employees["1050"])

# Q4. If an Employee ID does not exist, display Employee not found.
print("Q4:", employees.get("1050", "Employee not found"))

# Q5-Q7. Display salary of 1032, department of 1087, and email of 1015.
print("Q5 salary:", employees.get("1032", {}).get("Salary", "Employee not found"))
print("Q6 department:", employees.get("1087", {}).get("Department", "Employee not found"))
print("Q7 email:", employees.get("1015", {}).get("Email", "Employee not found"))

# Q8-Q10. Find missing emails, missing phones, and incomplete records.
missing_emails = [emp_id for emp_id, details in employees.items() if details.get("Email") is None]
missing_phones = [emp_id for emp_id, details in employees.items() if details.get("Phone") is None]
incomplete_records = [emp_id for emp_id, details in employees.items() if details.get("Email") is None or details.get("Phone") is None]
print("Q8 missing emails:", missing_emails)
print("Q9 missing phones:", missing_phones)
print("Q10 incomplete record count:", len(incomplete_records))

# Q11. Replace missing email addresses with Not Available (without changing source records).
employees_email_filled = {}
for emp_id, details in employees.items():
    record = details.copy()
    if record.get("Email") is None:
        record["Email"] = "Not Available"
    employees_email_filled[emp_id] = record
print("Q11 sample:", next((record for record in employees_email_filled.values() if record["Email"] == "Not Available"), "No missing emails"))

# Q12-Q14. Find IT employees, Finance employees, and employees managed by Manager_4.
for department in ("IT", "Finance"):
    print("Q department", department, [(emp_id, d["Name"]) for emp_id, d in employees.items() if d.get("Department") == department])
print("Q14 Manager_4:", [d["Name"] for d in employees.values() if d.get("Manager") == "Manager_4"])

# Q15-Q16. Find employees older than 50 and younger than 30.
print("Q15 older than 50:", [d["Name"] for d in employees.values() if d.get("Age", 0) > 50])
print("Q16 younger than 30:", [d["Name"] for d in employees.values() if d.get("Age", 0) < 30])

# Q17-Q19. Find highest/lowest paid employee and average salary.
salary_records = [(emp_id, d) for emp_id, d in employees.items() if isinstance(d.get("Salary"), (int, float))]
highest_paid = max(salary_records, key=lambda item: item[1]["Salary"])
lowest_paid = min(salary_records, key=lambda item: item[1]["Salary"])
print("Q17 highest salary:", highest_paid[0], highest_paid[1]["Name"], highest_paid[1]["Salary"])
print("Q18 lowest salary:", lowest_paid[0], lowest_paid[1]["Name"], lowest_paid[1]["Salary"])
print("Q19 average salary:", sum(d["Salary"] for _, d in salary_records) / len(salary_records))

# Q20-Q21. Display employees earning above 100,000 and below 50,000.
print("Q20 over 100000:", [(d["Name"], d["Salary"]) for d in employees.values() if d.get("Salary", 0) > 100000])
print("Q21 under 50000:", [(d["Name"], d["Salary"]) for d in employees.values() if d.get("Salary", 0) < 50000])

# Q22-Q23. Count employees by department and by manager.
from collections import Counter
department_counts = Counter(d.get("Department", "Unknown") for d in employees.values())
manager_counts = Counter(d.get("Manager", "Unknown") for d in employees.values())
print("Q22 department counts:", dict(department_counts))
print("Q23 manager counts:", dict(manager_counts))

# Q24. Find the department with the most employees.
print("Q24 largest department:", max(department_counts, key=department_counts.get), max(department_counts.values()), "employees")

# Q25. Increase each salary by 10% in a separate dictionary.
employees_with_raise = {}
for emp_id, details in employees.items():
    record = details.copy()
    if isinstance(record.get("Salary"), (int, float)):
        record["Salary"] = round(record["Salary"] * 1.10, 2)
    employees_with_raise[emp_id] = record
print("Q25 sample updated salary:", employees_with_raise["1001"]["Salary"])

# Q26. Update Employee 1040's department to Research in a copy.
employees_department_updated = {emp_id: details.copy() for emp_id, details in employees.items()}
if "1040" in employees_department_updated:
    employees_department_updated["1040"]["Department"] = "Research"
print("Q26:", employees_department_updated.get("1040", "Employee not found"))

# Q27. Add a new employee to a copy of the dictionary.
new_employees = {emp_id: details.copy() for emp_id, details in employees.items()}
new_id = str(max(int(emp_id) for emp_id in employees) + 1)
new_employees[new_id] = {"Name": "New Employee", "Department": "IT", "Age": 25, "Salary": 60000, "Email": "new.employee@company.com", "Phone": "9000000000", "Manager": "Manager_1"}
print("Q27 added:", new_id, new_employees[new_id])

# Q28. Delete Employee 1018 from a copy.
employees_without_1018 = {emp_id: details.copy() for emp_id, details in employees.items() if emp_id != "1018"}
print("Q28 employee 1018 remains:", "1018" in employees_without_1018)

# Q29. Update Employee 1075's phone number in a copy.
employees_phone_updated = {emp_id: details.copy() for emp_id, details in employees.items()}
if "1075" in employees_phone_updated:
    employees_phone_updated["1075"]["Phone"] = "9000000000"
print("Q29:", employees_phone_updated.get("1075", "Employee not found"))
```

```python
# Assignment questions: Employee JSON database, Q30-Q40
# Q30. Report total employees, missing email/phone counts, and salary statistics.
salaries = [d["Salary"] for d in employees.values() if isinstance(d.get("Salary"), (int, float))]
print("Q30 Employee report")
print("Total employees:", len(employees))
print("Missing email:", sum(d.get("Email") is None for d in employees.values()))
print("Missing phone:", sum(d.get("Phone") is None for d in employees.values()))
print("Average salary:", round(sum(salaries) / len(salaries), 2))
print("Highest salary:", max(salaries))
print("Lowest salary:", min(salaries))

# Q31. Generate a report listing employees department-wise.
print("Q31 Employees by department")
for department in sorted(department_counts):
    print("\\n" + department)
    for details in employees.values():
        if details.get("Department") == department:
            print(details["Name"])

# Q32. List employees by salary from highest to lowest.
print("Q32 Salary descending:")
for emp_id, details in sorted(employees.items(), key=lambda item: item[1].get("Salary", 0), reverse=True):
    print(emp_id, details["Name"], details.get("Salary"))

# Q33-Q34. Names containing 25; salaries between 60,000 and 80,000 inclusive.
print("Q33 Names containing 25:", [d["Name"] for d in employees.values() if "25" in d.get("Name", "")])
print("Q34 Salary between 60000 and 80000:", [(d["Name"], d["Salary"]) for d in employees.values() if 60000 <= d.get("Salary", 0) <= 80000])

# Q35. Find managers supervising more than 8 employees.
print("Q35 Managers with more than 8 employees:", {manager: count for manager, count in manager_counts.items() if count > 8})

# Q36. Find the department with the highest average salary.
department_salary_totals = {}
for details in employees.values():
    department = details.get("Department", "Unknown")
    department_salary_totals.setdefault(department, []).append(details["Salary"])
department_averages = {department: sum(values) / len(values) for department, values in department_salary_totals.items()}
print("Q36 Department highest average salary:", max(department_averages.items(), key=lambda item: item[1]))

# Q37-Q38. Employees missing both contact fields; duplicate non-missing phone numbers.
print("Q37 Missing both email and phone:", [d["Name"] for d in employees.values() if d.get("Email") is None and d.get("Phone") is None])
phone_numbers = [d["Phone"] for d in employees.values() if d.get("Phone") is not None]
phone_counts = Counter(phone_numbers)
print("Q38 Duplicate phone numbers:", {phone: count for phone, count in phone_counts.items() if count > 1})

# Q39-Q40. Employees with company.com emails; separate dictionary of IT employees.
print("Q39 company.com emails:", [d["Name"] for d in employees.values() if isinstance(d.get("Email"), str) and d["Email"].lower().endswith("@company.com")])
it_employees = {emp_id: details.copy() for emp_id, details in employees.items() if details.get("Department") == "IT"}
print("Q40 IT employee IDs:", list(it_employees))
```
