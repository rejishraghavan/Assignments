```python
# Question 9  

Create a dictionary of five products.

Example

    Laptop
    Mouse
    Keyboard
    Monitor
    Printer

Each product should contain

    Product Name
    Price
    Quantity

Write a program to

    Display all products.
    Print the product with the highest price.
    Calculate the total inventory value

(Formula:  Inventory Value = Price × Quantity)
```

```python
products = {
    "P1": {"Product Name": "Laptop", "Price": 75000.00, "Quantity": 100},
    "P2": {"Product Name": "Mouse", "Price": 200.00, "Quantity": 50},
    "P3": {"Product Name": "Keyboard", "Price": 2500.00, "Quantity": 35},
    "P4": {"Product Name": "Monitor", "Price": 15000.00, "Quantity": 25},
    "P5": {"Product Name": "Printer", "Price": 25000.00, "Quantity": 15}
}

for id, details in products.items():
    print(f"Product Name: {details['Product Name']} - Price : {details['Price']} - Quantity : {details['Quantity']}")

highest_priced_product = None
highest_price = 0.0
for id, details in products.items():
    if details["Price"] > highest_price:
        highest_price = details["Price"]
        highest_priced_product = details["Product Name"]
print(f"Highest Priced Product : {highest_priced_product}, Price : {highest_price}")

total_inventory_value = 0.0
for id, details in products.items():
    product_value = details["Price"] * details["Quantity"]
    total_inventory_value += product_value
print(f"Total Inventory value : {total_inventory_value}")
```

```python
# Question 10

Create a dictionary of five students and their marks.

Example

    Alice : 85
    Bob : 92
    Charlie : 78
    David : 88
    Emma : 95

Write a Python program to

    Find the student with the highest marks.
    Find the student with the lowest marks.
    Calculate the average marks.
    Display students who scored more than 85.
```

```python
student_marks = {
    "Alice": 85,
    "Bob": 92,
    "Charlie": 78,
    "David": 88,
    "Emma": 95
}

highest_mark_student = max(student_marks, key=student_marks.get)
lowest_mark_student = min(student_marks, key=student_marks.get)
print(f"Student with highest marks : {highest_mark_student}")
print(f"Student with lowest marks : {lowest_mark_student}")

total_marks = sum(student_marks.values())
num_students = len(student_marks)
average_marks = total_marks/num_students
print(f"Average marks : {average_marks}")
print(f"Students scored more than 85")
for student, marks in student_marks.items():
    if marks > 85:
        print(f"Name : {student}, Marks: {marks}")
```
