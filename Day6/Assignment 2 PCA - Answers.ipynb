```python
# Task 5: Create two strongly positively correlated features, calculate their covariance matrix, and interpret off-diagonal values.
# Answer: Off-diagonal covariance is +5, so the features increase together; these example features are perfectly correlated.
import numpy as np

feature_1 = np.array([1, 2, 3, 4, 5])
feature_2 = np.array([2, 4, 6, 8, 10])
covariance = np.cov(np.column_stack((feature_1, feature_2)), rowvar=False)
print("Covariance matrix:\n", covariance)
```

```python
# Task 7: For C = [[4, 2], [2, 3]], calculate eigenvalues/eigenvectors and find the leading eigenvector.
# Answer: Eigenvalues are about 5.5616 and 1.4384; PC1 is about [0.7882, 0.6154], up to sign.
import numpy as np

C = np.array([[4, 2], [2, 3]])
eigenvalues, eigenvectors = np.linalg.eigh(C)
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]
print("Eigenvalues, largest first:", eigenvalues)
print("Eigenvectors as columns:\n", eigenvectors)
print("First principal direction:", eigenvectors[:, 0])
```

```python
# Task 10: Implement PCA without sklearn: center, compute covariance/eigenpairs, sort, select components, and transform.
# Answer: The function below performs each PCA step and returns the transformed data and selected components.
import numpy as np

def pca(X, n_components):
    mean = X.mean(axis=0)
    centered = X - mean
    covariance = np.cov(centered, rowvar=False)
    eigenvalues, eigenvectors = np.linalg.eigh(covariance)
    order = np.argsort(eigenvalues)[::-1]
    components = eigenvectors[:, order[:n_components]]
    transformed = centered @ components
    return transformed, components, eigenvalues[order]

X = np.array([[2, 1], [3, 2], [4, 3], [5, 4], [6, 5]])
reduced, components, eigenvalues = pca(X, n_components=2)
print("Sorted eigenvalues:", eigenvalues)
print("Selected component vectors:\n", components)
print("Transformed data:\n", reduced)
```

```python
# Task 11: Project Question 10's centered data onto PC1 and calculate the 1D representation.
# Answer: Scores are approximately [-2.8284, -1.4142, 0, 1.4142, 2.8284], or the sign-reversed equivalent.
import numpy as np

X = np.array([[2, 1], [3, 2], [4, 3], [5, 4], [6, 5]])
centered = X - X.mean(axis=0)
covariance = np.cov(centered, rowvar=False)
eigenvalues, eigenvectors = np.linalg.eigh(covariance)
pc1 = eigenvectors[:, np.argmax(eigenvalues)]
one_dimensional = centered @ pc1
print("PC1:", pc1)
print("One-dimensional representation:", one_dimensional)
```

```python
# Task 12: Modify the PCA program so n_components is configurable and the reduced data is returned.
# Answer: This function selects the top n_components eigenvectors and returns an array with that many columns.
import numpy as np

def pca_reduce(X, n_components):
    mean = X.mean(axis=0)
    centered = X - mean
    covariance = np.cov(centered, rowvar=False)
    eigenvalues, eigenvectors = np.linalg.eigh(covariance)
    order = np.argsort(eigenvalues)[::-1]
    components = eigenvectors[:, order[:n_components]]
    return centered @ components

X = np.array([[2, 1], [3, 2], [4, 3], [5, 4], [6, 5]])
n_components = 1
reduced_data = pca_reduce(X, n_components)
print("Reduced data:\n", reduced_data)
print("Reduced shape:", reduced_data.shape)
```

```python
# Task 13: Use sklearn PCA to reduce the supplied 2D data to 1D; display original/transformed data and explained variance ratio.
# Answer: The transformed shape is (5, 1); the component explains 100% of variance because the features are collinear.
import numpy as np
from sklearn.decomposition import PCA

X = np.array([[2, 1], [3, 2], [4, 3], [5, 4], [6, 5]])
pca = PCA(n_components=1)
transformed = pca.fit_transform(X)
print("Original data:\n", X)
print("Transformed data:\n", transformed)
print("Explained variance ratio:", pca.explained_variance_ratio_)
```

```python
# Task 15: For Question 14's dataset, calculate component variance ratios, cumulative variance, and retention with 1/2 components.
# Answer: The first component retains essentially 100%; two components also retain essentially 100% (the second adds roundoff-level variance).
import numpy as np
from sklearn.decomposition import PCA

X = np.array([[2, 1, 5], [3, 2, 6], [4, 3, 7], [5, 4, 8], [6, 5, 9]])
pca = PCA(n_components=2).fit(X)
ratios = pca.explained_variance_ratio_
cumulative = np.cumsum(ratios)
print("Explained variance ratios:", ratios)
print("Cumulative explained variance:", cumulative)
print("Variance retained with 1 component:", cumulative[0])
print("Variance retained with 2 components:", cumulative[1])
```

## Optional End-to-End PCA Workflow

This supplementary workflow follows the standard eight-stage PCA process. The assignment provides synthetic arrays rather than a CSV/Excel file, and no target labels are supplied, so the example uses the Task 18 data in memory and an unsupervised clustering comparison instead of classification. It demonstrates preprocessing, variance analysis, visualization, reconstruction error, and downstream comparison.

```python
# 1. Setup and imports
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, mean_squared_error

# 2. Load and inspect the dataset
# The assignment supplies arrays, not a CSV/Excel file; use the Task 18 example.
data = np.array([[1, 100], [2, 200], [3, 300], [4, 400], [5, 500]], dtype=float)
feature_names = ["feature_1", "feature_2"]
df = pd.DataFrame(data, columns=feature_names)
print("Shape:", df.shape)
display(df.head())
print("Data types:\n", df.dtypes)
print("Missing values:\n", df.isna().sum())

# 3. Preprocess features for PCA
X = df[feature_names].to_numpy()
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Apply PCA with scikit-learn
pca_full = PCA().fit(X_scaled)
all_scores = pca_full.transform(X_scaled)
score_df = pd.DataFrame(all_scores, columns=[f"PC{i + 1}" for i in range(all_scores.shape[1])])
display(score_df)

# 5. Analyze explained variance
variance_ratio = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(variance_ratio)
n_components_95 = np.searchsorted(cumulative_variance, 0.95) + 1
pca_reduced = PCA(n_components=n_components_95)
reduced_scores = pca_reduced.fit_transform(X_scaled)
print("Explained variance ratio:", variance_ratio)
print("Cumulative explained variance:", cumulative_variance)
print("Components needed for at least 95%:", n_components_95)

# 6. Visualize PCA components
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
axes[0].bar(np.arange(1, len(variance_ratio) + 1), variance_ratio)
axes[0].set(title="Scree plot", xlabel="Principal component", ylabel="Explained variance ratio")
axes[1].plot(np.arange(1, len(cumulative_variance) + 1), cumulative_variance, marker="o")
axes[1].axhline(0.95, color="red", linestyle="--")
axes[1].set(title="Cumulative explained variance", xlabel="Number of components", ylabel="Cumulative ratio")
plt.tight_layout()
plt.show()

# The assignment has no class labels. Plot the first two components without class coloring.
pca_2d = PCA(n_components=2).fit_transform(X_scaled)
sns.scatterplot(x=pca_2d[:, 0], y=pca_2d[:, 1])
plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA projection (no class labels supplied)")
plt.show()

# 7. Reconstruct data and measure error
reconstructed_scaled = pca_reduced.inverse_transform(reduced_scores)
reconstruction_mse = mean_squared_error(X_scaled, reconstructed_scaled)
print("Reconstruction MSE in standardized feature space:", reconstruction_mse)

# 8. Compare downstream model performance before vs. after PCA
# With no target labels, compare unsupervised K-means silhouette scores and runtimes.
comparison = {}
for name, features in [("Standardized features", X_scaled), ("PCA features", reduced_scores)]:
    start = time.perf_counter()
    labels = KMeans(n_clusters=2, random_state=42, n_init=10).fit_predict(features)
    elapsed = time.perf_counter() - start
    comparison[name] = {"silhouette": silhouette_score(features, labels), "seconds": elapsed}
print(comparison)
print("No supervised train/test comparison is possible without target labels.")
```
