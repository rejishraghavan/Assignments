# A6: Functions and Modularity - Programming Answers

The following code answers Practical Questions 21-29 in order. Each solution uses built-in Python only; no NumPy or scikit-learn is required.

- **21.** `calculate_average(numbers)` returns the arithmetic mean; an empty list raises `ValueError`.
- **22.** `find_maximum(numbers)` scans values manually without `max()`; an empty list raises `ValueError`.
- **23.** The program separates input, mean, maximum, and display functions. `main()` reads user input when run as a script.
- **24.** `count_even(numbers)` counts integers divisible by 2.
- **25.** Accuracy is correct predictions divided by sample count; input lengths and empty input are validated.
- **26.** The function returns mean and population standard deviation (variance divided by $n$).
- **27.** Min-max normalization maps values to $[0, 1]$; constant input returns zeros to avoid division by zero.
- **28.** The pipeline loads sample records, removes rows with missing values, normalizes each feature, and displays results.
- **29.** Euclidean distance is computed from squared coordinate differences; point dimensions are checked.

```python
import math

# Question 21
def calculate_average(numbers):
    if not numbers:
        raise ValueError("numbers must not be empty")
    return sum(numbers) / len(numbers)

# Question 22: find the largest value without calling max().
def find_maximum(numbers):
    if not numbers:
        raise ValueError("numbers must not be empty")
    largest = numbers[0]
    for number in numbers[1:]:
        if number > largest:
            largest = number
    return largest

# Question 23
def read_numbers():
    return [float(value) for value in input("Enter numbers separated by spaces: ").split()]

def calculate_mean(numbers):
    return calculate_average(numbers)

def calculate_maximum(numbers):
    return find_maximum(numbers)

def display_results(mean, maximum):
    print("Mean:", mean)
    print("Maximum:", maximum)

def numbers_program():
    numbers = read_numbers()
    display_results(calculate_mean(numbers), calculate_maximum(numbers))

# Demonstrate the modular calculation without blocking for input.
sample_numbers = [2, 4, 6, 8]
display_results(calculate_mean(sample_numbers), calculate_maximum(sample_numbers))

# Question 24
def count_even(numbers):
    return sum(1 for number in numbers if number % 2 == 0)

print("Even count:", count_even([1, 2, 4, 7, 8]))

# Question 25
def calculate_accuracy(y_true, y_pred):
    if len(y_true) != len(y_pred):
        raise ValueError("y_true and y_pred must have equal lengths")
    if not y_true:
        raise ValueError("inputs must not be empty")
    correct = sum(actual == predicted for actual, predicted in zip(y_true, y_pred))
    return correct / len(y_true)

def display_result(accuracy):
    print("Accuracy: {:.2%}".format(accuracy))

display_result(calculate_accuracy([1, 0, 1, 1], [1, 0, 0, 1]))

# Question 26: population standard deviation.
def mean_and_standard_deviation(numbers):
    if not numbers:
        raise ValueError("numbers must not be empty")
    mean = sum(numbers) / len(numbers)
    variance = sum((value - mean) ** 2 for value in numbers) / len(numbers)
    return mean, math.sqrt(variance)

print("Mean and population standard deviation:", mean_and_standard_deviation([2, 4, 6, 8]))

# Question 27
def normalize(values):
    if not values:
        raise ValueError("values must not be empty")
    minimum = min(values)
    maximum = max(values)
    if maximum == minimum:
        return [0.0 for _ in values]
    return [(value - minimum) / (maximum - minimum) for value in values]

print("Normalized:", normalize([10, 20, 30, 40]))

# Question 28
def load_data():
    return [
        {"temperature": 20.0, "pressure": 100.0},
        {"temperature": 25.0, "pressure": 110.0},
        {"temperature": None, "pressure": 105.0},
        {"temperature": 30.0, "pressure": 120.0},
    ]

def remove_missing_values(rows):
    return [row for row in rows if all(value is not None for value in row.values())]

def normalize_data(rows):
    if not rows:
        return []
    columns = rows[0].keys()
    minima = {column: min(row[column] for row in rows) for column in columns}
    maxima = {column: max(row[column] for row in rows) for column in columns}
    normalized = []
    for row in rows:
        result = {}
        for column in columns:
            low, high = minima[column], maxima[column]
            result[column] = 0.0 if high == low else (row[column] - low) / (high - low)
        normalized.append(result)
    return normalized

def display_data(rows):
    for row in rows:
        print(row)

pipeline_data = remove_missing_values(load_data())
display_data(normalize_data(pipeline_data))

# Question 29
def euclidean_distance(point_a, point_b):
    if len(point_a) != len(point_b):
        raise ValueError("points must have the same number of coordinates")
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(point_a, point_b)))

print("Euclidean distance:", euclidean_distance([1, 2], [4, 6]))
```
