# A7: Execute a Calculator Program Through the Terminal

**Question 21:** Create `calculator.py` with separate `add()`, `subtract()`, `multiply()`, and `divide()` functions. Read two numbers and display all four results, without external libraries.

**Answer:** The code cell below is the complete program. Save it as `calculator.py`, then run `python calculator.py` (or `python3 calculator.py` if needed). Division by zero is reported rather than causing an exception.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return None
    return a / b

def main():
    first = float(input("Enter the first number: "))
    second = float(input("Enter the second number: "))
    print("Addition:", add(first, second))
    print("Subtraction:", subtract(first, second))
    print("Multiplication:", multiply(first, second))
    quotient = divide(first, second)
    if quotient is None:
        print("Division: undefined (cannot divide by zero)")
    else:
        print("Division:", quotient)

if __name__ == "__main__":
    main()
```
